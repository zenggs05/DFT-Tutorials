# Si band structure and DOS with py4vasp

Run this notebook in `VASP/examples/05-band-dos-si` after `job.slurm` has finished.
Requirements: VASP compiled with HDF5 (vaspout.h5), `pip install py4vasp`.
See Chapter 5 of the tutorial for explanations.

```python
import py4vasp

scf  = py4vasp.Calculation.from_path("1-scf")
band = py4vasp.Calculation.from_path("2-band-icharg11")
dos  = py4vasp.Calculation.from_path("3-dos")
opt  = py4vasp.Calculation.from_path("4-band-kpoints-opt")

# Fermi energy from the SCF run (the ICHARG=11 path run does not give a reliable E_F)
ef = scf.dos.read()["fermi_energy"]
print("E_F (SCF) =", ef, "eV")
```

## 1. Band structure from the two-step method (ICHARG=11)

```python
g = band.band.plot(fermi_energy=ef)
g.yrange = (-13, 6)
g
```

## 2. Fat bands (needs LORBIT = 11)

```python
g = band.band.plot("Si(s), Si(p)", fermi_energy=ef)
g.yrange = (-13, 6)
g
```

## 3. One-step method: KPOINTS_OPT (VASP >= 6.3)

```python
g = opt.band.plot("kpoints_opt")
g.yrange = (-13, 6)
g
```

```python
# band gap analysis written by VASP >= 6.4.3
print("fundamental gap:", opt.bandgap.fundamental())
print("direct gap     :", opt.bandgap.direct())
```

## 4. Density of states (dense mesh, ISMEAR = -5)

```python
g = dos.dos.plot("s, p")
g.xrange = (-13, 6)
g
```

## 5. Export data for Origin / Excel / your own matplotlib script

```python
band.band.to_csv(filename="band_icharg11.csv")
opt.band.to_csv("kpoints_opt", filename="band_kpoints_opt.csv")
dos.dos.to_csv(filename="dos.csv")
```
